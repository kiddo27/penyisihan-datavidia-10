### Mengatasi Null Values

```python
## Cek null values
print(f"null value ispu 2023: \n{ispu_2023.isna().sum()}\n")
print(f"null value ispu 2024: \n{ispu_2024.isna().sum()}\n")
print(f"null value ispu 2025: \n{ispu_2025.isna().sum()}\n")
```

```python
## Mengatasi null values ispu_2023
ispu_2023['parameter_pencemar_kritis'] = ispu_2023['parameter_pencemar_kritis'].fillna(ispu_2023['parameter_pencemar_kritis'].mode()[0])
ispu_2023.isna().sum()
```

```python
## Mengatasi null values ispu_2024
ispu_2024['pm_duakomalima'] = ispu_2024['pm_duakomalima'].fillna(ispu_2024['pm_duakomalima'].mean())
ispu_2024['pm_sepuluh'] = ispu_2024['pm_sepuluh'].fillna(ispu_2024['pm_sepuluh'].mean())
ispu_2024['sulfur_dioksida'] = ispu_2024['sulfur_dioksida'].fillna(ispu_2024['sulfur_dioksida'].mean())
ispu_2024['karbon_monoksida'] = ispu_2024['karbon_monoksida'].fillna(ispu_2024['karbon_monoksida'].mean())
ispu_2024['ozon'] = ispu_2024['ozon'].fillna(ispu_2024['ozon'].mean())
ispu_2024['nitrogen_dioksida'] = ispu_2024['nitrogen_dioksida'].fillna(ispu_2024['nitrogen_dioksida'].mean())
ispu_2024['max'] = ispu_2024['max'].fillna(ispu_2024['max'].mean())
ispu_2024['parameter_pencemar_kritis'] = ispu_2024['parameter_pencemar_kritis'].fillna(ispu_2024['parameter_pencemar_kritis'].mode()[0])


ispu_2024.isna().sum()
```

```python
## Mengatasi null values ispu_2025
ispu_2025['pm_duakomalima'] = ispu_2025['pm_duakomalima'].fillna(ispu_2025['pm_duakomalima'].mean())
ispu_2025['pm_sepuluh'] = ispu_2025['pm_sepuluh'].fillna(ispu_2025['pm_sepuluh'].mean())
ispu_2025['sulfur_dioksida'] = ispu_2025['sulfur_dioksida'].fillna(ispu_2025['sulfur_dioksida'].mean())
ispu_2025['karbon_monoksida'] = ispu_2025['karbon_monoksida'].fillna(ispu_2025['karbon_monoksida'].mean())
ispu_2025['ozon'] = ispu_2025['ozon'].fillna(ispu_2025['ozon'].mean())
ispu_2025['nitrogen_dioksida'] = ispu_2025['nitrogen_dioksida'].fillna(ispu_2025['nitrogen_dioksida'].mean())
ispu_2025['max'] = ispu_2025['max'].fillna(ispu_2025['max'].mean())
ispu_2025['parameter_pencemar_kritis'] = ispu_2025['parameter_pencemar_kritis'].fillna(ispu_2025['parameter_pencemar_kritis'].mode()[0])


ispu_2025.isna().sum()
```
